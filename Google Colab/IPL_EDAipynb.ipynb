**Visualizations**
**Winning Teams **

```python
matches["winner"].value_counts().plot(kind="bar")
plt.show()
```

**Runs Distribution**

```python
deliveries["batsman_runs"].hist()
plt.show()
```

**Toss Decision**

```python
matches["toss_decision"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.show()
```

**Top 10 Winning Teams**

```python
matches["winner"].value_counts().head(10).plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Winning Teams")
plt.xlabel("Team")
plt.ylabel("Number of Wins")
plt.xticks(rotation=45)
plt.show()
```

**Top 10 Run Scores**

```python
top_runs = deliveries.groupby("batter")["batsman_runs"].sum().sort_values(ascending=False).head(10)

top_runs.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Run Scorers")
plt.xlabel("Batter")
plt.ylabel("Runs")
plt.xticks(rotation=45)
plt.show()
```

**Top 10 Wicket Takers**

```python
top_wickets = deliveries.groupby("bowler")["is_wicket"].sum().sort_values(ascending=False).head(10)

top_wickets.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Wicket Takers")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)
plt.show()
```

**Top Six Hitters**

```python
sixes = deliveries[deliveries["batsman_runs"] == 6]

sixes.groupby("batter").size().sort_values(ascending=False).head(10).plot(kind="bar")

plt.title("Top Six Hitters")
plt.xlabel("Batter")
plt.ylabel("Sixes")
plt.xticks(rotation=45)
plt.show()
```

**Winning Margin**

```python
matches["result_margin"].plot(kind="hist", bins=20)

plt.title("Winning Margin Distribution")
plt.xlabel("Winning Margin")
plt.ylabel("Frequency")
plt.show()
```
